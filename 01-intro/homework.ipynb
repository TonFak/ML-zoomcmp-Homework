```python
XT = X.T
XT
```

```python
XTX = XT.dot(X)
```

```python
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
w = y.dot(X.dot(XTX_inv))
w
```
